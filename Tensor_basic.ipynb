# 기본 tensor 다루기


```python
import numpy as np
import torch
```

#### 1. list -> tensor

```python
data_list = [1, 2]
tensor = torch.tensor(data_list)
```

```python
print(f"tensor: {tensor}\n")
print(f"class type: {type(tensor)}\n")
print(f"data type: {tensor.dtype}\n")
print(f"dimension: {tensor.ndim}\n")
print(f"shape: {tensor.shape}") # tensor.size()와 같은 결과
```

#### 2. np array -> tensor

```python
data_np = np.array(
    [[1, 2],
    [3, 4]]
)
```

```python
# 방법 1 : tensor()
tensor_np = torch.tensor(data_np)  # data_np의 데이터를 복사해서 tensor를 생성.(메모리 공유 x)
tensor_np
```

```python
# 방법 2 : from_numpy()
tensor_from_numpy = torch.from_numpy(data_np) # data_np의 데이터를 가져와 tensor를 생성. (메모리 공유)
tensor_from_numpy
```

```python
# 차이점
data_np[0, 0] = -1
print(f"tensor():\n{tensor_np}\n")
print(f"from_numpy():\n{tensor_from_numpy}")
```

두 결과가 다른 것을 볼 수 있다.

#### 4. random값 tensor 생성
- 0과 1 사이의 랜덤한 값으로 tensor 생성.

```python
tensor1 = torch.rand(2, 3)
tensor2 = torch.rand([2, 3])
tensor3 = torch.rand((2, 3))
# 모두 같은 의미
```

```python
print(f"tensor1: {tensor1}\n")
print(f"tensor2: {tensor2}\n")
print(f"tensor3: {tensor3}\n")
```

```python
# 만약 0과 10 사이의 랜덤한 값으로 tensor를 생성하려면
min_val = 0
max_val = 10
rand_tensor = min_val + (max_val - min_val) * torch.rand(2, 3)
rand_tensor
```

- avg가 0, std가 1인 random값 tensor 생성.

```python
# avg가 0, std가 1인 random tensor 생성.
tensor = torch.randn(2, 3)
tensor
```

```python
# 만약 avg가 5, std가 2인 random tensor를 생성하려면
```

```python
avg = 5
std = 2
tensor = avg + 2 * torch.randn(2, 3)
tensor
```

#### 6. tensor -> np array 변환

```python
t_tensor = torch.ones(2, 2)
t_tensor
```

```python
# 첫 번째 방법 : numpy()
np_array = t_tensor.numpy() # t_tensor와 메모리를 공유.
np_array
```

```python
# 두 번째 방법 : np.array(tensor)
np_array2 = np.array(t_tensor) # t_tensor와 메모리를 공유하지 않음.
np_array2
```

```python
t_tensor[0, 0] = -1
```

```python
# 차이점
print(f"numpy():\n{np_array}\n")
print(f"np.array():\n{np_array2}\n")
```

# 기본 tensor 연산

```python
a = torch.tensor(
    [[1, 2],
     [3, 4]]
)
b = torch.tensor(
    [[1, 1],
     [2, 2]]
)
```

```python
print(a + b)
print(torch.add(a, b))
```

```python
print(a * b)
print(torch.multiply(a, b))
```

```python
# 행렬 곱은 두 행렬의 데이터 타입이 일치해야만 연산 가능.
print(torch.matmul(a, b))
print(a @ b)
```

```python
# 데이터 타입 변경
print(f"old: {a.dtype}")
a_new = a.to(torch.float32)
print(f"new: {a_new.dtype}")
```
